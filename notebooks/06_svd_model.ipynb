### Train SVD model

```python
import pandas as pd

from surprise import SVD, Dataset, Reader, dump
```

```python
ratings_data = pd.read_parquet("../data/movielens_ratings_w_tmdbId.parquet")
```

```python
ratings_data.shape
```

#### Write SVD model using ratings 32M data and save it to incorporate into hybrid_model.py

```python
reader = Reader(rating_scale=(0.5, 5.0))
data = Dataset.load_from_df(ratings_data[["userId", "tmdbId", "rating"]], reader)
svd = SVD(n_factors=100, n_epochs=30, lr_all=0.01, reg_all=0.2)
svd.fit(data.build_full_trainset())
```

```python
dump.dump("../models/svd.pkl", algo=svd)
```
